## 1. Datos de entrada: features por cliente

| Feature | Tipo | Por qué |

| `n_vehiculos`, `currently_connected_vehicles` | contador | Tamaño de la relación con la marca (particular vs. multi-vehículo vs. flota) |
| `visitas_taller_por_vehiculo` | contador (derivada) | Uso de taller en garantía, normalizado por nº de vehículos → target de planes de mantenimiento |
| `edad_cliente` | numérica (derivada) | Punto medio de `age_range`: conserva el orden de los tramos |
| `months_as_mb_customer` | numérica | Antigüedad como cliente → fidelización |

| `pct_electrificados` | numérica (derivada) | % de vehículos eléctricos/híbridos → afinidad por electrificación |
| `pct_premium` | numérica (derivada) | % de vehículos de gama alta (S, E, GLE/GLS, G, EQE/EQS, AMG, Maybach…) |

**No entran en el modelo** (se usan solo para describir los clusters): `gender_derived` (dato inferido; meterlo produciría segmentos por

```python
X_df = P.client_features(df)
print(f"Clientes: {len(X_df):,}  |  features de entrada: {len(P.INPUT_FEATURES)}  |  nulos en entrada: {int(X_df[P.INPUT_FEATURES].isna().sum().sum())}")
display(X_df[P.INPUT_FEATURES].describe(include="all").T)
```

## 2. Pipeline de transformación

K-Means usa distancias euclídeas, así que todas las variables tienen que estar en una escala comparable:

* **Contadores** (`n_vehiculos`, `currently_connected_vehicles`, `visitas_taller_por_vehiculo`): las flotas llegan a 561 vehículos. Se **winsorizan
  en el P99** (aprendido en `fit`), se aplica `log1p` y se estandarizan; sin esto, unas pocas flotas formarían sus propios clusters.
* **Numéricas**: `StandardScaler` (media 0, desviación 1).
* **Binaria**: se deja en 0/1.
* **Categóricas**: `OneHotEncoder` (no tienen orden).

Todo va en un `ColumnTransformer` de scikit-learn, reutilizable para transformar clientes nuevos con los mismos parámetros.

```python
pre = P.build_preprocessor()
X = pre.fit_transform(X_df)
cols_X = pre.get_feature_names_out()
print(f"Matriz de entrada al modelo: {X.shape[0]:,} x {X.shape[1]}")
display(pd.DataFrame(X, columns=cols_X).describe().T[["mean", "std", "min", "max"]])
pre
```

## 3. Ejecución de K-Means para un rango de k

Se entrena un K-Means para cada **k entre 2 y 10**: menos de 2 no es una segmentación y más de 10 segmentos no son manejables por un equipo
de marketing. Para cada k:

* `n_init=10`: se lanza 10 veces con centroides iniciales distintos (`k-means++`) y se queda la de menor inercia, para no depender de una mala inicialización.
* `random_state=42`: resultados reproducibles.

Se guardan los 9 modelos ajustados (`modelos[k]`) para reutilizarlos en el método del codo, el Silhouette Score y el modelo final.

```python
K_RANGE = range(2, 11)
modelos = P.run_kmeans_range(X, K_RANGE, n_init=10, random_state=42)
resumen_k = P.runs_summary(modelos)
display(resumen_k.style.format({"inercia": "{:,.0f}", "cluster menor (%)": "{:.1f}", "cluster mayor (%)": "{:.1f}", "tiempo (s)": "{:.1f}"}))
```

```python
# Tamaño de cada cluster (% de clientes) para cada k
tam = pd.DataFrame({k: pd.Series(np.bincount(m.labels_) / len(m.labels_) * 100) for k, m in modelos.items()}).T
tam.index.name = "k"; tam.columns = [f"c{i}" for i in tam.columns]
fig, ax = plt.subplots(figsize=(10, 4))
bottom = np.zeros(len(tam))
for c in tam.columns:
    v = tam[c].fillna(0).to_numpy()
    ax.bar(tam.index, v, bottom=bottom, edgecolor="white", linewidth=.6)
    bottom += v
ax.set_xlabel("k"); ax.set_ylabel("% de clientes"); ax.set_xticks(list(K_RANGE))
ax.set_title("Reparto de clientes entre clusters para cada k")
plt.tight_layout(); plt.show()
```

**Lectura:** la inercia (suma de distancias al cuadrado de cada cliente a su centroide) **siempre baja al aumentar k**, así que por sí sola no
sirve para elegir k; la usaremos en el **método del codo** para ver a partir de qué k deja de bajar de forma apreciable, y lo contrastaremos con el
**Silhouette Score**. La tabla también muestra si algún k genera clusters residuales (muy pequeños), que serían poco accionables para marketing.
